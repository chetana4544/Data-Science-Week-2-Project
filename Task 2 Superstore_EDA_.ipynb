# Task 2: Exploratory Data Analysis 

Here I performs EDA on the cleaned Superstore dataset using beginner-friendly Python code.


## 1. Import libraries and load the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\chetana kusekar\Downloads\superstore_cleaned.csv")

print("Dataset shape:", df.shape)
df.head()
```

### Simple explanation
- `pandas` helps us read and work with data.
- `matplotlib` and `seaborn` help us create graphs.
- `df.shape` tells us the number of rows and columns.
- `df.head()` displays the first five rows.



## 2. Check the dataset

```python
# Check column names and data types
df.info()

# Check missing values
print(df.isnull().sum())

# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

### Simple explanation
- `df.info()` shows column names, data types and how many values each column contains.
- `df.isnull().sum()` counts missing values in each column.
- `df.duplicated().sum()` counts rows that are exact duplicates
 To check the dataset's structure and make sure it is ready for analysis.

## 3. Descriptive statistics

```python
# Statistics for numeric columns
df.describe()
```

### Simple explanation
`df.describe()` gives a statistical summary of numerical columns.

- **count:** number of available values
- **mean:** average
- **std:** how spread out the values are
- **min:** smallest value
- **25%:** first quartile
- **50%:** median (middle value)
- **75%:** third quartile
- **max:** largest value

 To understand the basic numbers in our dataset, such as sales, profit and quantity.

## 4. Prepare category and region columns

```python
# The cleaned file stores category and region as separate 0/1 columns.
# This code creates simple readable category and region columns.

category_columns = [col for col in df.columns if col.startswith("category_")]
region_columns = [col for col in df.columns if col.startswith("region_")]

df["Category"] = df[category_columns].idxmax(axis=1).str.replace("category_", "", regex=False)
df["Region"] = df[region_columns].idxmax(axis=1).str.replace("region_", "", regex=False)

df[["Category", "Region"]].head()
```


 The cleaned file stores category and region as separate 0/1 columns. These new columns make grouping and charting easier.

## 5. Histogram and KDE plot

```python
# Distribution of Sales
sns.histplot(data=df, x="sales", kde=True, color="skyblue")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Count")
plt.show()
```


 To see whether sales values are mostly low, medium or high, and how they are spread out.

## 6. Correlation matrix

```python
# Check relationship between important numeric columns
correlation = df[["sales", "quantity", "profit"]].corr()

print(correlation)

sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```


- The code selects the sales, quantity and profit columns.
- `.corr()` calculates how strongly these numerical columns move together.
- `sns.heatmap()` displays the results as a coloured chart.
- `annot=True` prints the correlation numbers inside the chart.

 A value near `+1` means a strong positive relationship, near `-1` means a strong negative relationship, and near `0` means little or no linear relationship.
 To check whether sales, quantity and profit are related. Correlation does not prove that one causes another.

## 7. Box plot

```python
# Compare sales in each category
sns.boxplot(data=df, x="Category", y="sales")
plt.title("Sales by Category")
plt.xticks(rotation=15)
plt.show()
```


 To compare the spread and middle value of sales across categories and spot possible outliers.

## 8. Bar chart

```python
# Total sales for each category
category_sales = df.groupby("Category")["sales"].sum()

category_sales.plot(kind="bar", color="orange")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```


 To compare total sales across product categories and identify the highest and lowest.

## 9. Monthly sales trend

```python
# Convert order_date into date format
df["order_date"] = pd.to_datetime(df["order_date"])

# Calculate monthly sales
monthly_sales = df.groupby(df["order_date"].dt.to_period("M"))["sales"].sum()

monthly_sales.plot(kind="line", marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```


To see how sales change over time and identify months with higher or lower sales.

## 10. Scatter plot

```python
# Check relationship between sales and profit
sns.scatterplot(data=df, x="sales", y="profit", hue="Category")
plt.title("Sales vs Profit")
plt.show()
```


 To see whether higher sales tend to occur with higher profit and to notice unusual records.

## 11. Grouped analysis

```python
# Sales and profit by category
category_summary = df.groupby("Category").agg({
    "sales": "sum",
    "profit": "sum",
    "quantity": "sum"
})

print(category_summary)

# Sales by region
region_summary = df.groupby("Region")["sales"].sum()
print(region_summary)
```


 To compare category and regional performance in one summary.

## 12. Important insights

```python
# Find highest and lowest sales categories
print("Highest sales category:", category_sales.idxmax())
print("Lowest sales category:", category_sales.idxmin())

# Find highest and lowest profit categories
profit_by_category = df.groupby("Category")["profit"].sum()
print("Highest profit category:", profit_by_category.idxmax())
print("Lowest profit category:", profit_by_category.idxmin())

# Sales and profit correlation
print("Sales-Profit Correlation:", df["sales"].corr(df["profit"]))
```

### Simple explanation
- `idxmax()` returns the category name with the highest value.
- `idxmin()` returns the category name with the lowest value.
- `groupby()` calculates total profit for each category.
- `.corr()` calculates the relationship between sales and profit.

**Why do we do this?** To quickly identify important findings from the dataset without checking every row manually.
